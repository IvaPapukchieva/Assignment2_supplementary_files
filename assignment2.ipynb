```python
# populate the graph 
g = Graph()
# load the schema 
g.parse("Bookshop.ttl", format="ttl")
print(g.serialize(format="ttl"))

```

```python
BOOKSHOP = Namespace("http://example.org/bookshop#")
SCHEMA = Namespace("https://schema.org/")
WEBSHOP = Namespace("http://example.org/webshop")

g.bind("bookshop", BOOKSHOP)
g.bind("schema", SCHEMA)
g.bind("webshop", WEBSHOP)
g.bind("rdfs", RDFS)
g.bind("owl", OWL)


```

```python
df = pd.read_csv("group_10_ken3140_webshop.csv").dropna()
print(f"Removed {1000-len(df)} rows with empty cells.")

selected_columns = [
    "Item Name",
    "Item URL",
    "Item URI",
    "Category",
    "Subcategory",
    "Publisher",
    "Attribute 1",
    "Value 1",
    "Schema URI 1",
    "Attribute 2",
    "Value 2",
    "Schema URI 2",
    "Attribute 3",
    "Value 3",
    "Schema URI 3",
    "Attribute 4",
    "Value 4",
    "URI 4",
    "Attribute 5",
    "Value 5",
    "URI 5"
]
df 
```

```python
def create_uri(input_string):
  """
  Creates a URI from an input string
  """
  import re
  cleaned_string = re.sub(r'[^\w]', '', input_string)
  uri_string = BOOKSHOP[cleaned_string]
  return uri_string

test_uri_1 = create_uri("The Godfather: Part II")
test_uri_2 = create_uri("Francis Ford Coppola")
print(test_uri_1, test_uri_2)
```

```python
for index, row in df.iterrows():

    # Individual product
    item_uri = URIRef(row["Item URI"])

    # Product's subcategory
    subcategory_uri = BOOKSHOP[row["Subcategory"]]
    g.add((item_uri, RDF.type, subcategory_uri))

    #Product's category
    category_uri = SCHEMA[row["Category"]]
    g.add((item_uri, RDF.type, category_uri))

    # Product URL
    g.add((item_uri, SCHEMA.url, URIRef(row["Item URL"])))

    # Publisher (using create_uri helper)
    publisher_name = row["Publisher"].strip()
    publisher_uri = create_uri(publisher_name)

    g.add((publisher_uri, RDF.type, BOOKSHOP.Publisher))
    g.add((item_uri, BOOKSHOP.publishedBy, publisher_uri))

    # Five attributes
    for i in range(1, 6):

        value = row[f"Value {i}"]

        if i <= 3:
            property_uri = URIRef(row[f"Schema URI {i}"])
        else:
            property_uri = URIRef(row[f"URI {i}"])

        #proper formating of the properties
        if property_uri == SCHEMA.price:
            value_literal = Literal(value, datatype=XSD.decimal)
        elif property_uri == SCHEMA.numberOfPages:
            value_literal = Literal(value, datatype=XSD.integer)
        elif property_uri == SCHEMA.datePublished:
            value_literal = Literal(value, datatype=XSD.date)
        else:
            value_literal = Literal(value)

        g.add((item_uri, property_uri, value_literal))

print(g.serialize(format="ttl"))
```

```python
#saving the latest version of the graph
g.serialize(destination="group_10_ken3140_webshop.ttl", format="turtle")
```

# Provide a query and the resulting response using the query function of RDFLib on your graph for each of the following:

a) For a given item (select a random item from your RDF Graph), provide all its
categories and subcategories, and its brand.

```python
webshop = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX bookshop: <http://example.org/bookshop#>

SELECT ?category ?subcategory ?publisher
WHERE {

    <http://example.org/webshop/the-girls-9781529970982>
        rdf:type ?subcategory .

    ?subcategory
        rdfs:subClassOf ?category .

    <http://example.org/webshop/the-girls-9781529970982>
        bookshop:publishedBy ?publisher .
}
"""

for row in g.query(webshop):
    print(row.category, row.subcategory, row.publisher)
```

b) Provide items from different subcategories that have the same brand.

```python
query_b = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX bookshop: <http://example.org/bookshop#>
PREFIX schema: <https://schema.org/>

SELECT DISTINCT ?name1 ?subcategory1 ?name2 ?subcategory2 ?publisher
WHERE {
    ?item1 schema:name ?name1 ;
            bookshop:publishedBy ?publisher ;
            rdf:type ?subcategory1 .

    ?item2 schema:name ?name2 ;
            bookshop:publishedBy ?publisher ;
            rdf:type ?subcategory2 .

        FILTER(?item1 != ?item2)
        FILTER(?subcategory1 != ?subcategory2)

        FILTER(?subcategory1 != schema:Book)
        FILTER(?subcategory2 != schema:Book)
}
"""
for row in g.query(query_b):
    print(row.name1, row.subcategory1, row.name2, row.subcategory2, row.publisher)
```

c) Group products by brand and show the average price or rating for each brand.

```python
query_c = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX bookshop: <http://example.org/bookshop#>
PREFIX schema: <https://schema.org/>

SELECT ?publisher (AVG(?price) as ?averagePrice)
WHERE {
    ?item bookshop:publishedBy ?publisher ;
            schema:price ?price .

}
GROUP BY ?publisher
ORDER BY DESC(?averagePrice)
"""
for row in g.query(query_c):
    print(row.publisher, row.averagePrice)
```

d) Sort products in one category according to average brand price or rating.

```python
query_d = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX bookshop: <http://example.org/bookshop#>
PREFIX schema: <https://schema.org/>

SELECT ?name ?publisher ?averagePublisherPrice
WHERE {
    ?item rdf:type bookshop:Romance ;
            schema:name ?name ;
            bookshop:publishedBy ?publisher ;
    {
        SELECT ?publisher (AVG(?price) as ?averagePublisherPrice)
        WHERE {
            ?product bookshop:publishedBy ?publisher ;
                    schema:price ?price .
        }
        GROUP BY ?publisher
    }
}
ORDER BY DESC(?averagePublisherPrice)
"""
for row in g.query(query_d):
    print(row.name, row.publisher, row.averagePublisherPrice)
```

```python
import time
from SPARQLWrapper import SPARQLWrapper, JSON

def get_wikidata_facts(brand_name):

    endpoint = SPARQLWrapper("https://qlever.dev/api/wikidata") #use qlever
    endpoint.setReturnFormat(JSON)
    endpoint.addCustomHttpHeader("User-Agent", "KEN3140-Assignment2/1.0")

    query_e = f"""
    PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
    PREFIX skos: <http://www.w3.org/2004/02/skos/core#>
    PREFIX wdt: <http://www.wikidata.org/prop/direct/>

    SELECT ?companyLabel ?headquartersLabel ?founderLabel ?countryLabel ?inceptionDate ?website
    WHERE {{

      {{ ?company rdfs:label "{brand_name}"@en . }}
      UNION
      {{ ?company skos:altLabel "{brand_name}"@en . }}
      BIND("{brand_name}" AS ?companyLabel)
      OPTIONAL {{ ?company wdt:P159 ?hq . ?hq rdfs:label ?headquartersLabel . FILTER(lang(?headquartersLabel) = "en") }}

      OPTIONAL {{ ?company wdt:P112 ?f . ?f rdfs:label ?founderLabel . FILTER(lang(?founderLabel) = "en") }}

      OPTIONAL {{ ?company wdt:P17 ?c . ?c rdfs:label ?countryLabel . FILTER(lang(?countryLabel) = "en") }}

      OPTIONAL {{ ?company wdt:P571 ?inceptionDate . }}

      OPTIONAL {{ ?company wdt:P856 ?website . }}
    }}

    LIMIT 1
    """

    endpoint.setQuery(query_e)

    return endpoint.query().convert()["results"]["bindings"]



top_brand = "HarperCollins"

facts = get_wikidata_facts(top_brand)

for row in facts:
    for key, val in row.items():
        print(f"{key}: {val['value']}")
```

```python
target_item = URIRef("http://example.org/webshop/the-deal-9780349447155") # Book THe Deal

check_subcategory_query = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX bookshop: <http://example.org/bookshop#>

SELECT ?subcategory
WHERE {
    ?targetItem rdf:type ?subcategory .
    ?subcategory rdfs:subClassOf+ bookshop:Genre .
}
"""

for row in g.query(check_subcategory_query, initBindings={"targetItem": target_item}):
    print(row.subcategory)
```

```python
query_f = """
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX schema: <https://schema.org/>
PREFIX bookshop: <http://example.org/bookshop#>

SELECT ?recommendedItem ?recommendedItemName ?publisher (IF(?publisher = ?targetPublisher, 1, 0) AS ?samePublisher)
WHERE {
    ?targetItem rdf:type ?subcategory ;
                bookshop:publishedBy ?targetPublisher .
    ?subcategory rdfs:subClassOf+ bookshop:Genre .

    ?recommendedItem rdf:type ?subcategory ;
                      schema:name ?recommendedItemName ;
                      bookshop:publishedBy ?publisher .
    FILTER (?recommendedItem != ?targetItem)
}
ORDER BY DESC(?samePublisher)
LIMIT 5
"""

for row in g.query(query_f, initBindings={"targetItem": target_item}):
    print(row.recommendedItem, row.recommendedItemName, row.publisher, row.samePublisher)
```

- Which items are priced above the average price of their own subcategory?


```python
query_g = """

PREFIX bookshop: <http://example.org/bookshop#>
PREFIX schema: <https://schema.org/>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>



SELECT ?itemName ?subcategory ?itemPrice ?subcategoryAveragePrice

WHERE {

    ?item rdf:type ?subcategory ;
          schema:name ?itemName ;
          schema:price ?itemPrice .
    ?subcategory rdfs:subClassOf+ bookshop:Genre .

    {
        SELECT ?subcategory (AVG(?price) AS ?subcategoryAveragePrice)

        WHERE {
            ?otherItem rdf:type ?subcategory ;
                       schema:price ?price .
            ?subcategory rdfs:subClassOf+ bookshop:Genre .
        }

        GROUP BY ?subcategory
    }

    FILTER (?itemPrice > ?subcategoryAveragePrice)
}

ORDER BY ?subcategory DESC(?itemPrice)
"""

for row in g.query(query_g):
    print(row.itemName, "-", row.subcategory, "- price:", row.itemPrice, "- category avg:", row.subcategoryAveragePrice)
```
